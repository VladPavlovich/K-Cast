```python
val = {"Premise A": "",
"Premise B": "",
"Conclusion": "",
"Validity": ""
}

import json
```

```python
with open("val.json", "r") as f:
    data = json.load(f)
```

```python
final_json = {}
```

```python
for data_dict in data:
    k = data_dict['id']
    v = data_dict['syllogism']
    val_data = v
    val_a = val_data.split('.')[0]
    val_b = val_data.split('.')[1]
    val_c = val_data.split('.')[2]
    v_write = {"Premise A": val_a,
                    "Premise B": val_b,
                    "Conclusion": val_c,
                    "Validity": data_dict['validity']}
    final_json[k] = v_write
```

```python
final_json
with open("val_parsed.json", "w") as f:
    json.dump(final_json, f, indent=4)
```
